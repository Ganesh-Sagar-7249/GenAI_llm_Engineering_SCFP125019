```python
clubs = [
    "Coding Club",
    "AI & Machine Learning Club",
    "Robotics Club",
    "Photography Club",
    "Sports Club",
    "Music Club",
    "Drama Club",
    "Dance Club",
    "Literary Club",
    "Entrepreneurship Club",
    "Environmental Club"
]
```

```python
print("Available clubs:")
for club in clubs:
    print("-", club)
```

```python
def recommend_club(student_interest):

    messages = [
        {
            "role": "system",
            "content": f"""
You are a college club recommendation assistant.

The available college clubs are:

{clubs}

The student will describe their interests using free text.

Your task is to recommend exactly ONE club from the available clubs.

Give the response in exactly this format:

Recommended Club: <club name>
Reason: <one-line reason>

Do not recommend a club that is not in the list.
"""
        },
        {
            "role": "user",
            "content": student_interest
        }
    ]

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=messages
    )

    return response.choices[0].message.content
```

```python
print(recommend_club("I love programming, Python, AI and building software."))
```
